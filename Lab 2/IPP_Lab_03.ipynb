```python
import cv2
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np
```

```python
img_cv = cv2.imread('sample.jpg')
img_pil = Image.open('sample.jpg')
```

```python
plt.imshow(cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB))
plt.title('Original Image')
plt.axis('off')
plt.show()
```

```python
cropped = img_cv[100:300, 150:400]
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.title('Cropped Image (OpenCV)')
plt.axis('off')
plt.show()
```

```python
box = (150, 100, 400, 300)  # (left, upper, right, lower)
cropped_pil = img_pil.crop(box)
cropped_pil.show()
```

```python
resized = cv2.resize(img_cv, (300, 200))  # width, height
plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.title('Resized Image (OpenCV)')
plt.axis('off')
plt.show()
```

```python
resized_pil = img_pil.resize((300, 200))
resized_pil.show()
```

```python
(h, w) = img_cv.shape[:2]
center = (w//2, h//2)
M = cv2.getRotationMatrix2D(center, 45, 1.0)
rotated = cv2.warpAffine(img_cv, M, (w, h))
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title('Rotated by 45° (OpenCV)')
plt.axis('off')
plt.show()
```

```python
rotated_pil = img_pil.rotate(45)
rotated_pil.show()
```

```python
flip_horizontal = cv2.flip(img_cv, 1)
flip_vertical = cv2.flip(img_cv, 0)
```

```python
titles = ['Original', 'Horizontal Flip', 'Vertical Flip']
images = [img_cv, flip_horizontal, flip_vertical]
plt.figure(figsize=(10,3))
for i in range(3):
    plt.subplot(1,3,i+1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB))
    plt.title(titles[i])
    plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(img_cv, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title('Grayscale')
plt.axis('off')
plt.show()
```

```python
hsv = cv2.cvtColor(img_cv, cv2.COLOR_BGR2HSV)
plt.imshow(cv2.cvtColor(hsv, cv2.COLOR_HSV2RGB))
plt.title('HSV Color Space')
plt.axis('off')
plt.show()
```

```python
rgb_converted = cv2.cvtColor(hsv, cv2.COLOR_HSV2BGR)
```

```python
h, s, v = cv2.split(hsv)
plt.figure(figsize=(12,3))
for i, (img, title) in enumerate(zip([h, s, v], ['Hue', 'Saturation', 'Value'])):
    plt.subplot(1,3,i+1)
    plt.imshow(img, cmap='gray')
    plt.title(title)
    plt.axis('off')
plt.show()
```

```python
from PIL import Image

img = Image.open('flower.jpg')
img.show()
# Crop (left, upper, right, lower)
crop_img = img.crop((100, 100, 400, 400))
crop_img.show()
# Resize
resize_img = img.resize((200, 200))
resize_img.show()

```

```python
import cv2
img = cv2.imread('flower.jpg')
# Rotate 90 degrees clockwise
rotated = cv2.rotate(img, cv2.ROTATE_90_CLOCKWISE)
# Flip horizontally
flipped = cv2.flip(img, 1)
# Display results
cv2.imshow('Rotated', rotated)
cv2.imshow('Flipped', flipped)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Convert between color spaces
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
# Show all
cv2.imshow('Gray', gray)
cv2.imshow('HSV', hsv)
cv2.imshow('RGB', rgb)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
