```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from PIL import Image
```

```python
pil_img = Image.open("flower.jpg")
plt.imshow(pil_img)
plt.title("Image Loaded Using PIL")
plt.axis('off')
plt.show()
```

```python
cv_img = cv2.imread("flower.jpg")
cv_img_rgb = cv2.cvtColor(cv_img, cv2.COLOR_BGR2RGB)
plt.imshow(cv_img_rgb)
plt.title("Image Loaded Using OpenCV (RGB)")
plt.axis('off')
plt.show()
```

```python
gray = cv2.cvtColor(cv_img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
plt.show()
```

```python
hist_manual, bins = np.histogram(gray.flatten(), 256, [0, 256])
plt.plot(hist_manual, color='black')
plt.title("Manual Histogram (NumPy)")
plt.xlabel("Intensity Values")
plt.ylabel("Frequency")
plt.show()

```

```python
plt.hist(gray.flatten(), bins=256, color='gray')
```

```python
hist_cv = cv2.calcHist([gray], [0], None, [256], [0, 256])
```

```python
r_min = np.min(gray)
r_max = np.max(gray)
```

```python
equalized = cv2.equalizeHist(gray)
```

```python
stretched = ((gray - r_min) * (255 / (r_max - r_min))).astype(np.uint8)
```

```python
clahe = cv2.createCLAHE(clipLimit=3.0, tileGridSize=(8,8))
clahe_img = clahe.apply(gray)
```

```python
titles = ["Original", "Stretched", "Equalized", "CLAHE"]
images = [gray, stretched, equalized, clahe_img]

```
