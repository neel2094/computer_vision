```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
```

```python
img = cv2.imread("flower.jpg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)   # convert BGR → RGB for display

plt.imshow(img)
plt.title("Original Image")
plt.axis('off')
plt.show()
```

```python
gaussian_noise = np.random.normal(0, 25, img.shape).astype(np.int16)
noisy_gaussian = img.astype(np.int16) + gaussian_noise
noisy_gaussian = np.clip(noisy_gaussian, 0, 255).astype(np.uint8)
plt.imshow(noisy_gaussian)
plt.title("Gaussian Noise Added")
plt.axis('off')
plt.show()
```

```python
sp = img.copy()
prob = 0.02  # noise probability
# Salt
num_salt = int(prob * img.size * 0.5)
coords = [np.random.randint(0, i - 1, num_salt) for i in img.shape[:2]]
sp[coords[0], coords[1]] = 255
# Pepper
num_pepper = int(prob * img.size * 0.5)
coords = [np.random.randint(0, i - 1, num_pepper) for i in img.shape[:2]]
sp[coords[0], coords[1]] = 0
plt.imshow(sp)
plt.title("Salt & Pepper Noise Added")
plt.axis('off')
plt.show()
```

```python
avg_3 = cv2.blur(img, (3,3))
avg_7 = cv2.blur(img, (7,7))
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(avg_3)
plt.title("Averaging 3×3")
plt.axis('off')
plt.subplot(1,2,2)
plt.imshow(avg_7)
plt.title("Averaging 7×7")
plt.axis('off')
plt.show()
```
